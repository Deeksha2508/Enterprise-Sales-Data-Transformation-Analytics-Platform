```python
import pandas as pd
import numpy as np

df = pd.read_csv("superstore.csv", encoding="cp1252")

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

display(df.head())
```

```python
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values)
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate Records:", duplicate_count)
```

```python
print(df.dtypes)
```

```python
#Convert dates

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    errors="coerce"
)
```

```python
#Check invalid dates
print("Invalid Order Dates:",
      df["Order Date"].isnull().sum())

print("Invalid Ship Dates:",
      df["Ship Date"].isnull().sum())
```

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Records before:", before)
print("Records after:", after)
print("Duplicates removed:", before - after)
```

```python
numeric_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

for col in numeric_columns:
    print(
        col,
        "Min =", df[col].min(),
        "Max =", df[col].max()
    )
```

```python
#Create year
df["Year"] = df["Order Date"].dt.year
#create month
df["Month"] = df["Order Date"].dt.month
#create month name
df["Month Name"] = df["Order Date"].dt.month_name()
```

```python
df["Profit Margin"] = np.where(
    df["Sales"] != 0,
    (df["Profit"] / df["Sales"]) * 100,
    0
)
```

```python
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
# Round Sales and Profit
df["Sales"] = df["Sales"].round(2)
df["Profit"] = df["Profit"].round(2)

# Check final Python total
print("Python Total Sales:", df["Sales"].sum())

# Check number of rows
print("Total Records:", len(df))

# Check that Shipping Days exists
print("Shipping Days column exists:", "Shipping Days" in df.columns)

# Export final cleaned dataset
df.to_csv(
    "cleaned_superstore.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
print("File: cleaned_superstore.csv")
```

```python
quality_report = {
    "Total Records": len(df),
    "Total Columns": len(df.columns),
    "Duplicate Records": df.duplicated().sum(),
    "Missing Values": df.isnull().sum().sum(),
    "Invalid Order Dates": df["Order Date"].isnull().sum(),
    "Invalid Ship Dates": df["Ship Date"].isnull().sum()
}

for key, value in quality_report.items():
    print(f"{key}: {value}")
```
